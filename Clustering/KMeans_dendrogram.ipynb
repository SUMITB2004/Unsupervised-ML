```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()

X = iris.data
y = iris.target
```

```python
# scaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Visualize

sns.scatterplot(x=X_scaled[:,0], y=X_scaled[:,2], c=y)
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram

# Linkage matrix
z = linkage(X_scaled, method="ward")

# Plot
plt.figure(figsize=(12,6))
dendrogram(z)
plt.xlabel("samples")
plt.ylabel("distance")
plt.title("dendrogram")
```

```python
# clustering
from sklearn.cluster import AgglomerativeClustering

agg = AgglomerativeClustering(n_clusters = 2)       #so according to scatter plot(dendrogram) above we are gonna use k=2
labels = agg.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0], y=X_scaled[:,2], c=labels)
```
